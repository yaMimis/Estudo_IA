## IMPORTS SKLEARN ##

```python
import numpy as np
from sklearn.datasets import load_wine
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.linear_model import SGDClassifier, LinearRegression, RidgeClassifier, ElasticNet, BayesianRidge
from sklearn.model_selection import KFold, StratifiedKFold
from sklearn.metrics import confusion_matrix, f1_score, accuracy_score, root_mean_squared_error, mean_squared_error, ConfusionMatrixDisplay
```

## CARREGAR DB ##

```python
dataset = load_wine()

x = dataset.data
y = dataset.target
```

```python
dataset = load_diabetes()

x = dataset.data
y = dataset.target
```

## DADOS PARA TREINO E TESTE ##

```python
x_treino, x_teste, y_treino, y_teste = train_test_split(x, y, test_size=0.2)
```

## MODELOS DE TESTE##

# MODELO CLASSIFICAÇÃO

```python
model = svm.SVC()
model.fit(x_treino, y_treino)
```

```python
model = SGDClassifier()
model.fit(x_treino, y_treino)
```

```python
model = RidgeClassifier()
model.fit(x_treino, y_treino)
```

# MODELO REGRESSÃO

```python
model = ElasticNet()
model.fit(x_treino, y_treino)
```

```python
model = BayesianRidge()
model.fit(x_treino, y_treino)
```

```python
model = LinearRegression()
model.fit(x_treino, y_treino)
```

```python
result = model.predict(x_teste)
```

```python
print(result)
```

```python
print(y_teste)
```

## METRICAS PARA ANALISE DE RESULTADO

# METRICAS P/CLASSIFICAÇÃO

```python
print(accuracy_score(y_teste, result))
print(f1_score(y_teste, result, average="micro"))
print(confusion_matrix(y_teste, result))
```

# METRICAS P/REGRESSÃO

```python
print(mean_squared_error(y_teste, result))
print(root_mean_squared_error(y_teste, result))
```

```python
kf = StratifiedKFold(n_splits=5, shuffle=True)
model = svm.SVC()
acc = []
f1score = []
for i, (train_index, test_index) in enumerate(kf.split(x, y)):
    model.fit(x[train_index], y[train_index])
    result = model.predict(x[test_index])
    acc_f = accuracy_score(y[test_index], result)
    f1s_f = f1_score(y[test_index], result, average="macro")
    acc.append(acc_f)
    f1score.append(f1s_f)

print(f"média acc: {np.mean(acc)}")
print(f"média f1_score: {np.mean(f1score)}")
```
